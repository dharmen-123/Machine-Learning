```python
X=df.drop(columns=['price'],axis=1)
y=df['price']
```

```python
X
```

```python
df.columns
```

```python
X_one_encode=pd.get_dummies(X,columns=['model','transmission','fuelType'],drop_first=True )
```

```python
X_one_encode
```

```python
X_one_encode=X_one_encode.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder
columns=['model','transmission','fuelType']
Xlable=X.copy()
label_encoders={}

for i in columns:
    le=LabelEncoder()
    Xlable[i]=le.fit_transform(Xlable[i].astype(str))
    label_encoders[i]=le
```

```python
Xlable
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_col=['year','mileage','tax','mpg', 'engineSize']
scaler= StandardScaler()
X_one_encode[numerical_col]=scaler.fit_transform(X_one_encode[numerical_col])
```

```python
X_one_encode
```

```python
Xlable[Xlable.columns]=scaler.fit_transform(Xlable[Xlable.columns])
```

```python
Xlable
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
```

#### Model using One hot Encoding data

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encode,y,train_size=0.33,random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
y_test
```

```python
r2= r2_score(y_test,y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1))/(n-p-1)
print("Adjusted R2 Score : ",adjusted_r2)
```

#### Model using Label Ecoding Data

```python
X_train, X_test, y_train, y_test = train_test_split(Xlable,y,train_size=0.33,random_state=42)
```
