### ************ Linear Regression *******************

Linear Regression is a Supervised Machine Learning algorithm used to predict a continuous numerical value by finding the best-fit straight line between the input features and the target variable.

Simple Definition

Linear Regression predicts a numerical value by fitting the best straight line through the data points.

* It assumes that there is a linear relationship between the input and output
* Uses a best‑fit line to make predictions
* Commonly used in forecasting, trend analysis, and predictive modelling

* Independent variable (input): Hours studied because it's the factor we control or observe.
* Dependent variable (output): Exam score because it depends on how many hours were studied.

Linear regression uses the independent variable to predict the dependent variable.

* Best Fit Line

In linear regression, the best-fit line is the straight line that best represents the relationship between the independent variable (input) and the dependent variable (output). The goal is to minimize the difference between the actual data points and the predicted values generated by the model.

* Goal of the Best-Fit Line

The goal of linear regression is to find a straight line that minimizes the error (the difference) between the observed data points and the predicted values. This line helps us predict the dependent variable for new, unseen data.

* Equation of the Best-Fit Line

For simple linear regression (with one independent variable), the best-fit line is represented by the equation

        y=mx+b

* Where:

* y is the predicted value (dependent variable)
* x is the input (independent variable)
* m is the slope of the line (how much y changes when x changes)
* b is the intercept (the value of y when x = 0)

The best-fit line will be the one that optimizes the values of m (slope) and b (intercept) so that the predicted y values are as close as possible to the actual data points.

Implementation

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [3, 4, 5, 6, 7, 8]
}

df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[["Experience"]]
y = df["Salary"]

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

# Predictions on Training Data
y_pred = model.predict(X)


# Predict Salary for 7 Years Experience
prediction = model.predict([[7]])

print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("Predicted Salary:", prediction[0])
```

```python
# Performance metrices 
from sklearn.metrics import r2_score

# R² Score
r2 = r2_score(y, y_pred)

# Adjusted R²
n = len(y)          # Number of observations
p = X.shape[1]      # Number of independent variables

adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))
print("\nPerformance Metrics")
print("-------------------")
print("R² Score:", r2)
print("Adjusted R²:", adjusted_r2)
```

#### Overfitting & Underfitting

#### Overfitting
* Overfitting occurs when the model learns the training data too well, including its noise and outliers.

Characteristics
* Very low training error
* High testing error
* Excellent performance on training data
* Poor performance on unseen data


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X = np.array([[1],[2],[3],[4],[5],[6]])
y = np.array([3,4,5,6,7,8])

poly = PolynomialFeatures(degree=5)

X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly,y)

x_range = np.linspace(1,6,100).reshape(-1,1)

plt.scatter(X,y,color="blue")

plt.plot(
    x_range,
    model.predict(poly.transform(x_range)),
    color="red"
)

plt.title("Overfitting Example")
plt.show()
```

#### Underfitting
* Underfitting occurs when the model is too simple to learn the relationship between the input and output.

Characteristics
* High training error
* High testing error
* Poor predictions
* Model cannot capture the pattern

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1,2,3,4,5,6])
y = np.array([3,4,5,6,7,8])

prediction = np.repeat(5.5,len(X))

plt.scatter(X,y,color='blue')
plt.plot(X,prediction,color='red',linewidth=3)
plt.title("Underfitting")
plt.show()
```
