## *********** K Nearest Neighbour ***********

K-Nearest Neighbors (KNN) is a simple yet powerful machine learning algorithm used for both classification and regression. It predicts outcomes by looking at the K closest data points and deciding based on majority vote (classification) or average (regression). The key is choosing the right value of K and distance metric.

* Type of Algorithm: Non-parametric, instance-based, lazy learner (no training phase, computation happens at prediction time).

#### How It Works:
* Choose a value of K (number of neighbors).
* Compute the distance between the query point and all training points (commonly Euclidean, Manhattan, Minkowski, cosine similarity).
* Select the K nearest neighbors.
* For classification → majority vote; for regression → average value.



##### Distance Metrics:
* Euclidean: straight-line distance.
* Manhattan: sum of absolute differences.
* Cosine similarity: useful for text/embeddings.
* Hamming: for categorical/binary data.

##### Choosing K:
* Small K → sensitive to noise, risk of overfitting.
* Large K → smoother decision boundary, risk of underfitting.
* Often chosen using cross-validation or elbow method.
