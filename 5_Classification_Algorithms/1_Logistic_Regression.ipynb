### *************** Logistic Regression ****************

* Logistic Regression is a Supervised Machine Learning Classification algorithm used to predict the probability that an input belongs to a particular class.

* Although its name contains "Regression", it is not a regression algorithm. It is primarily used for classification problems, especially Binary Classification.

Definition

* Logistic Regression is a supervised learning algorithm that predicts the probability of an observation belonging to a particular class using the Sigmoid (Logistic) function.

#### Types of Logistic Regression

A. Binary Logistic Regression
* Predicts only two classes.

Examples:
* Spam / Not Spam
* Pass / Fail
* Yes / No
* Fraud / Not Fraud

B. Multinomial Logistic Regression
* Predicts more than two classes.

Examples:

* Cat
* Dog
* Horse

C. Ordinal Logistic Regression

* Used when classes have an order.

Examples:

* Poor
* Average
* Good
* Excellent

```python
#  Implementation
import pandas as pd
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings("ignore")

# Sample Dataset
data = {
    "Hours": [1, 2, 3, 4, 5, 6],
    "Pass":  [0, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Pass"]

# Create Model
model = LogisticRegression()

# Train Model
model.fit(X, y)

# Predict Class
prediction = model.predict([[4.5]])

# Predict Probability
probability = model.predict_proba([[4.5]])

print("Predicted Class:", prediction[0])
print("Prediction Probability:", probability)
```

#### Sigmoid Function
* The Sigmoid function converts any real number into a probability between 0 and 1.

Formula
* P=1/(1+e^-z)

where
* P = Probability
* e = Euler's Number (≈ 2.718)
* z = Linear equation

* It measures how far the predicted probabilities are from the actual class labels.
* Lower Log Loss indicates a better model. A Log Loss of 0 represents perfect predictions.
