## *************** Desicion Tree *************** 

Decision Trees are supervised learning models used for both classification and regression. They split data into subsets based on feature values, forming a tree-like structure of decisions. Their biggest strengths are interpretability and flexibility, but they can overfit if not pruned properly.

Structure:

* Root Node → represents the entire dataset, split on the best feature.
* Internal Nodes → represent feature tests.
* Branches → represent outcomes of tests.
* Leaf Nodes → represent final predictions (class label or value).


Splitting Criteria:

* Entropy & Information Gain (ID3) → measures reduction in uncertainty.
* Gain Ratio (C4.5) → adjusts for bias toward many-valued attributes.
* Gini Index (CART) → measures impurity; lower values are better.

Stopping Conditions:

* Maximum depth reached.
* Minimum samples per split.
* No further information gain.

Advantages:
* Easy to interpret and visualize.
* Handles both categorical and numerical data.
* Requires little preprocessing.

Disadvantages:
* Prone to overfitting.
* Sensitive to small data changes.
* Biased toward features with many levels.

Solutions:

* Pruning (pre-pruning or post-pruning).
* Use ensemble methods like Random Forests or Gradient Boosted Trees.

Example
Problem: Predict whether a customer will buy a product.

* Root Node: Income > ₹50,000?
*    If No → No Purchase.
*    If Yes → check Age > 30?
        * If No → No Purchase.
        * If Yes → check Previous Purchases?
            * If Yes → Purchase.
            * If No → No Purchase.
* This flowchart-like decision path leads to a prediction.
