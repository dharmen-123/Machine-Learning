## ****** Model Evaluation in Classification ******

Model evaluation is the process of checking how well a classification machine learning model performs on unseen/test data.

* A classification model predicts a class/category, such as:
* Spam / Not Spam
* Disease / No Disease
* Pass / Fail
* Cat / Dog
* Fraud / Not Fraud

The main purpose of evaluation is to determine:

* How accurate the model's predictions are.
* How well the model generalizes to unseen data.
* What types of mistakes the model makes.
* Whether the model is suitable for the real-world application.

#### A model can perform very well on training data but poorly on new data. This is called overfitting.
* Therefore, we generally divide the dataset into:
### Training Data → Model learns
### Testing Data → Model is evaluated

Example:
Dataset
   |
*  +---- Training Data (80%) → Train Model
   |
*  +---- Testing Data (20%)  → Evaluate Model

## 4. Recall
Recall is also called:
* Sensitivity
* True Positive Rate (TPR)
  
Recall tells us:
* Out of all actual positive cases, how many did the model correctly identify?

* Example
  


* Interpretation
* The model detected 80% of all actual disease cases.

When is Recall important?
* Recall is important when False Negatives are costly.

## 5.F1-Score
* F1-score is the harmonic mean of Precision and Recall.
* It provides a single metric that balances both.




    
When to use F1?
*  F1-score is particularly useful when:
* Classes are imbalanced.
* Both Precision and Recall are important.
* You want a single metric balancing Precision and Recall.
