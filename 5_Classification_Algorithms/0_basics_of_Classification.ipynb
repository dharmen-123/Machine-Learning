### **************** Classification ************************

* Classification is a Supervised Machine Learning algorithm used to predict categorical (discrete) output values by learning from labeled training data.

* Instead of predicting a continuous value (like Regression), Classification predicts the class or category to which an input belongs.

Definition

* Classification is a supervised learning technique that assigns input data into one of the predefined classes or categories.

### Types of Classification

#### A) Binary Classification
Predicts only two classes.

Examples:
* Spam / Not Spam
* Yes / No
* Pass / Fail
* Fraud / Not Fraud
* Disease / No Disease

#### B) Multi-Class Classification
Predicts more than two classes.

Examples:
* Cat
* Dog
* Horse
* Bird
*     OR
* Red
* Green
* Blue


#### C) Multi-Label Classification
One input can belong to multiple classes simultaneously.

Example
Movie Genres
* Input:
* Movie

Output:
* Action ✓
* Comedy ✓
* Adventure ✓
* A movie can have multiple labels.
