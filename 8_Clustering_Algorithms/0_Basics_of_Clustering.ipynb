### **************** Clustering *****************

Clustering is an unsupervised machine learning technique used to group similar data points into clusters such that data points within the same cluster are more similar to each other than to those in other clusters.

#### Applications of Clustering

* Customer Segmentation
* Recommendation Systems
* Image Segmentation
* Fraud Detection
* Document Classification

### Types of Clustering Algorithms
* 1. K-Means Clustering
* 2. Hierarchical Clustering
* 3. DBSCAN
* 4. Mean Shift Clustering
* 5. Gaussian Mixture Model (GMM)
