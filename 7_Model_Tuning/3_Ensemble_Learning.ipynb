#### **************** Ensemble Learning Techniques  ***********************

Ensemble Learning is a machine learning technique in which multiple models (called weak learners or base learners) are combined to create a stronger model that achieves better accuracy and generalization than any single model.

#### Why Use Ensemble Learning?

Suppose you have three models:
* Model A Accuracy = 80%
* Model B Accuracy = 82%
* Model C Accuracy = 85%
* Instead of relying on one model, combine all three models.
    * Result:
    * Ensemble Accuracy = 90%+
