#### ***************  HYPERPARAMETER TUNING ******************

In Machine Learning, a hyperparameter is a parameter whose value is set before training the model and is not learned from the data.

Hyperparameter Tuning is the process of selecting the optimal hyperparameter values of a machine learning algorithm to improve model performance and generalization on unseen data.

#### Why are Hyperparameters Important?
Hyperparameters control:

* Model complexity
* Training speed
* Accuracy
* Overfitting and Underfitting

Bad hyperparameters can make even a good algorithm perform poorly.

#### Methods of Hyperparameter Tuning
1. Grid Search
2. Random Search
3. Bayesian Optimization
4. Cross-Validation Based Tuning
