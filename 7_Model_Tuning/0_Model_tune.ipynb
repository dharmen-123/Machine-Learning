### **************  MODEL TUNING *****************

Model Tuning (Hyperparameter Tuning) is the process of finding the best set of hyperparameters for a machine learning model to improve its performance, accuracy, and generalization on unseen data.

Model Tuning is the process of optimizing a machine learning model by adjusting its hyperparameters to achieve the best performance while minimizing overfitting and underfitting.

#### What are Hyperparameters?

Hyperparameters are settings that are defined before training the model.

| Algorithm      | Hyperparameter                    |
| -------------- | --------------------------------- |
| KNN            | K (number of neighbors)           |
| Decision Tree  | max\_depth, min\_samples\_split   |
| Random Forest  | n\_estimators                     |
| SVM            | C, Gamma, Kernel                  |
| Neural Network | Learning Rate, Epochs, Batch Size |


#### Why is Model Tuning Necessary?

* A model may suffer from:
1. Underfitting
* Model learns too little from data.
*Poor performance on training and testing data.

Example:

*KNN with K = 50 for a small dataset.

2. Overfitting
* Model memorizes training data.
* Performs well on training but poorly on testing data.

Example:

* Decision Tree with unlimited depth.

3. Good Fit
* Learns actual patterns.
* Generalizes well on unseen data.

Example: KNN Model Tuning
* Suppose we want to predict whether a tumor is malignant or benign.

* Without Tuning
    * knn = KNeighborsClassifier(n_neighbors=1)

Result:
* High variance
* Overfitting

* After Tuning
    * knn = KNeighborsClassifier(n_neighbors=5)

Result:
* Better generalization
* 8Improved accuracy
